### GRAPHICAL INSIGHTS FROM DATA

#### 1. TOTAL SALES BY CITY

```python
df1=df.copy()
df1

df1["Total_Sales"]=(df1["Units Sold"]*df1["Price Per Unit"]) 
```

```python
sales_by_city=df1.groupby('City')["Total_Sales"].sum().reset_index()
sales_by_city
```

```python
plt.figure(figsize=(8, 5))
plt.bar(sales_by_city['City'], sales_by_city['Total_Sales'], color='skyblue', edgecolor='black')

# Add labels and title
plt.xlabel('City', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.title('Total Sales by City', fontsize=14)
plt.xticks(rotation=40)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

#### INSIGHTS:
#### Delhi has highest total sales while Vadodara has least total sales.

#### 2. TOTAL QUANTITY BY MONTH

```python
quantity_by_month=df1.groupby('Month')["Units Sold"].sum().reset_index()
quantity_by_month
```

```python
from matplotlib.ticker import MultipleLocator


plt.figure(figsize=(8, 5))
plt.plot(quantity_by_month['Month'], quantity_by_month['Units Sold'], marker='o',linestyle='-',linewidth='2',markersize='6',
     color='skyblue',label=quantity_by_month['Units Sold'])

for x, y in zip(quantity_by_month['Month'],quantity_by_month['Units Sold']):
    plt.text(x, y + 5, str(y), ha='center', fontsize=9)

# Add labels and title
plt.xlabel('Month', fontsize=12)
plt.ylabel('Units Sold', fontsize=12)
plt.title('Total Quantity by Month', fontsize=14)
plt.xlim(0,13)
plt.gca().xaxis.set_major_locator(MultipleLocator(1))


# Show the plot
plt.tight_layout()
plt.show()
```

#### INSIGHTS:
#### July month has highest total quantity sold while February has least total quantity sold. 

#### 3. TOTAL SALES BY DAYNAME

```python
sales_by_dayname=df1.groupby('Day Name')["Total_Sales"].sum().reset_index()
sales_by_dayname
```

```python
plt.figure(figsize=(8, 5))
plt.fill_between(sales_by_dayname['Day Name'], sales_by_dayname['Total_Sales'], color="lightblue", alpha=0.4)
plt.plot(sales_by_dayname['Day Name'],sales_by_dayname['Total_Sales'] , color="Slateblue", alpha=0.8)

# Add labels and title
plt.xlabel('Day Name', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.title('Total Sales by Day Name', fontsize=14)
plt.xticks(rotation=45)


# Show the plot
plt.tight_layout()
plt.show()
```

#### INSIGHTS:
#### Generally, total sales is highest on Saturdays and least on Wednesdays.

#### 4. COUNT OF CUSTOMER RATINGS

```python
import seaborn as sns

# Ensure ratings are treated as categorical for proper ordering
df1['ratings'] = pd.Categorical(df1['Customer Ratings'], categories=[1, 2, 3, 4, 5], ordered=True)

# Plot count of ratings
plt.subplots(figsize=(6, 4))
sns.countplot(data=df1, x='ratings', palette='viridis',label='data')

# Add labels and title
plt.xlabel("Customer Rating")
plt.ylabel("Count")
plt.title("Count of Customer Ratings")

# Show plot
plt.show()
```

#### INSIGHTS:
#### High number of customers like the products as they have given 5-star ratings. While, the few customers have also given 1-star ratings. So, the product quality and services can be improved to attract more 5-star ratings.

#### 5. Transactions by Payment Method

```python
payment_counts = df1.groupby("Payment Method")["Transaction ID"].count()

# -------------------------------
# Plot Pie Chart with Labels
# -------------------------------
plt.figure(figsize=(6, 6))
colors = plt.cm.Paired.colors  # Nice color palette

# autopct shows percentage, also include count in label
def autopct_format(pct, all_vals):
    absolute = int(round(pct/100.0 * sum(all_vals)))
    return f"{pct:.1f}%\n({absolute})"

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct=lambda pct: autopct_format(pct, payment_counts),
    startangle=90,
    colors=colors
)

plt.title("Transactions by Payment Method", fontsize=14)
plt.tight_layout()
plt.show()
```

#### INSIGHTS:
#### Majority of people using UPI payment method, which means UPI services and its availability need to be improved. People are attracting towards real time UPI transactions more.

#### 6. TOTAL SALES BY MOBILE MODEL

```python
sales_by_mobile_model=df1.groupby('Mobile Model')["Total_Sales"].sum().reset_index()
top_sales_by_mobile_model= sales_by_mobile_model.sort_values(by="Total_Sales",ascending=False).head(5)
top_sales_by_mobile_model
```

```python
plt.figure(figsize=(8, 5))
plt.bar(top_sales_by_mobile_model['Mobile Model'],top_sales_by_mobile_model['Total_Sales'], color=plt.cm.Paired.colors, edgecolor='black')

# Add labels and title
plt.xlabel('Mobile Model', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.title('Mobile Models With Highest Total Sales', fontsize=14)
plt.xticks(rotation=45)

# Show the plot
plt.tight_layout()
plt.show()
```

#### INSIGHTS:
#### iPhone SE has highest sales followed by OnePlus Nord, Galaxy Note 20.

```python
sales_by_mobile_model=df1.groupby('Mobile Model')["Total_Sales"].sum().reset_index()
least_sales_by_mobile_model= sales_by_mobile_model.sort_values(by="Total_Sales",ascending=False).tail(5)
least_sales_by_mobile_model
```

```python
plt.figure(figsize=(8, 5))
plt.bar(least_sales_by_mobile_model['Mobile Model'],least_sales_by_mobile_model['Total_Sales'], 
        color=plt.cm.viridis(np.linspace(0, 1, 5)), edgecolor='black')

# Add labels and title
plt.xlabel('Mobile Model', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.title('Mobile Models With Highest Total Sales', fontsize=14)
plt.xticks(rotation=45)

# Show the plot
plt.tight_layout()
plt.show()
```

#### INSIGHTS:
#### Redmi 9 has least total sales followed by Vivo S1, OnePlus 9.
